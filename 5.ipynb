```python
def divisores(n):
    lista=[]
    for i in range (1,n+1):
        if n%i==0:
            lista.append(i)
    return lista

if __name__=='__main__':
    n=int(input("Introduzca un número positivo: "))
    listaDivisores = divisores(n)
    print(listaDivisores)
```

```python
def union(lista1, lista2):
    resultado = list(lista1) 
    for elemento in lista2:
        if elemento not in resultado:
            resultado.append(elemento)
    return resultado

def interseccion(lista1, lista2):
    resultado = []
    for elemento in lista1:
        if elemento in lista2 and elemento not in resultado:
            resultado.append(elemento)
    return resultado

if __name__ == "__main__":
    l1 = [1, 2, 3, 4, 5]
    l2 = [4, 5, 6, 7, 8]
    print(f"Lista 1: {l1}")
    print(f"Lista 2: {l2}")
    print(f"Unión: {union(l1, l2)}")
    print(f"Intersección: {interseccion(l1, l2)}")
```

```python
def validarDNI(dni):
    letras_validas = "TRWAGMYFPDXBNJZSQVHLCKE"
    
    if len(dni) != 9:
        return False
    
    parte_num = dni[:8]
    letra_usuario = dni[8].upper()
    
    if not parte_num.isdigit() or not letra_usuario.isalpha():
        return False
        
    numero = int(parte_num)
    resto = numero % 23
    letra_correcta = letras_validas[resto]
    
    return letra_usuario == letra_correcta

if __name__ == "__main__":
    dni_input = input("Introduce un DNI (8 números y 1 letra): ")
    if validarDNI(dni_input):
        print("El DNI es VÁLIDO.")
    else:
        print("El DNI es FALSO.")
```

```python
def esLuhnValido(numero_str):
    lista = [int(d) for d in numero_str]
    
    lista.reverse()
    
    suma_total = 0
    
    for i in range(len(lista)):
        valor = lista[i]
        
        if i % 2 != 0:
            valor = valor * 2
            
        suma_digitos = (valor // 10) + (valor % 10)
        
        suma_total += suma_digitos
        
    return (suma_total % 10) == 0

if __name__ == "__main__":
    tarjeta = input("Introduce número de tarjeta o IMEI para validar: ")
    tarjeta = tarjeta.replace(" ", "")
    
    if tarjeta.isdigit():
        if esLuhnValido(tarjeta):
            print("El número es VÁLIDO (Cumple Luhn).")
        else:
            print("El número NO es válido.")
    else:
        print("Error: Introduce solo números.")
```
