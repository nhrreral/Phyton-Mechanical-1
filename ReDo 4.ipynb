```python
from math import sqrt

def esPrimo(n):
    pri=True
    for i in range (2,int(sqrt(n)+1)):
        if n%i==0: 
            pri=False
    return pri

if __name__ == '__main__':
    nm=int(input("Introduzca un número positivo: "))
    if esPrimo(nm)==True:
        print("El numero es", nm, "Y", nm+1)   
    else:
        print("El número no es primo")

    
    
```

```python
from math import pi
from math import pow
from math import exp

def gaussiana(x):
    fx = (1/pow(2*pi,0.5))*exp(-((x**2)/2))
    return fx

def trapecios(a,b,n):
    h=(b-a)/n
    suma=0
    for k in range (1,n):
        xk=a+k*h
        suma+=gaussiana(xk)
    A=h/2*(gaussiana(a)+2*suma+gaussiana(b))
    return A

if __name__ == '__main__':
    a=int(input("Introduzca un primer número: "))   
    b=int(input("Introduzca un segundo número: "))   
    n=int(input("Introduzca un tercer número: "))
    Area = trapecios(a,b,n)
    print(Area)
```

```python
def fibonacci(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    elif n>1:
        return fibonacci(n-1)+fibonacci(n-2)
    else:
        print("Error")
        
if __name__ == '__main__':
    n=int(input("Introduzca un primer número: "))
    fib=fibonacci(n)
    print(fib)
```
