```python
ACELERACION_MOTOR = 20.0
VELOCIDAD_LIMITE = 1000.0
GRAVEDAD = -9.8

tiempo = 0
velocidad = 0.0
altitud = 0.0
motor_encendido = True
tiempo_apagado = 0


while velocidad >= 0:
    
    print(f"T: {tiempo}s | Vel: {velocidad:.1f} m/s | Alt: {altitud:.1f} m")
    
    altitud = altitud + velocidad
    if motor_encendido:
        velocidad = velocidad + ACELERACION_MOTOR
        
        if velocidad >= VELOCIDAD_LIMITE:
            velocidad = VELOCIDAD_LIMITE
            motor_encendido = False
            tiempo_apagado = tiempo + 1
            print(f"--- MOTOR APAGADO (T={tiempo_apagado}s) ---")
    else:
        velocidad = velocidad + GRAVEDAD

    tiempo = tiempo + 1

print(f"Momento de apagado del motor: {tiempo_apagado} s")
print(f"Tiempo total de ascenso (aprox.): {tiempo - 1} s")
print(f"Altitud máxima alcanzada (aprox.): {altitud:.1f} m")
```

```python
capital = float(input("Importe total del préstamo (€): "))
tin_anual = float(input("Interés anual: "))
plazo_meses = int(input("Plazo en meses: "))

interes_mensual = (tin_anual / 100) / 12
capital_pendiente = capital
total_intereses = 0.0

if interes_mensual == 0:
    cuota = capital / plazo_meses
else:
    cuota = capital * (interes_mensual * (1 + interes_mensual)**plazo_meses) / (((1 + interes_mensual)**plazo_meses) - 1)

print(f"\nCuota mensual fija: {cuota:.2f} €\n")

print(f"{'Mes':>5} | {'Intereses':>12} | {'Amortización':>12} | {'Pendiente':>15}")
print(f"{'-'*5} | {'-'*12} | {'-'*12} | {'-'*15}")

for mes in range(1, plazo_meses + 1):
    
    intereses = capital_pendiente * interes_mensual
    
    amortizacion = cuota - intereses
    
    capital_pendiente = capital_pendiente - amortizacion
    
    total_intereses = total_intereses + intereses
    
    print(f"{mes:>5} | {intereses:12.2f} € | {amortizacion:12.2f} € | {capital_pendiente:15.2f} €")

total_pagado = cuota * plazo_meses
print("\n--- RESUMEN FINAL ---")
print(f"Total pagado: {total_pagado:.2f} €")
print(f"Total de intereses: {total_intereses:.2f} €")
```
